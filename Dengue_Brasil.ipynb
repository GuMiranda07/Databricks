## 0. Importando Bibliotecas

```python
#manipulação de dados
import numpy as np
import pandas as pd

# visualização de dados
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px 

#Base de Dados
df_dengue = pd.read_excel('Base de Dados Dengue.xlsx')
```

```python
#Informações sobre a base

df_dengue.info()
```

## 2. Verificando valores das tabelas



```python
#Total de valores únicos por coluna
df_dengue.nunique()
```

```python
#Analisando os primeiros registros
df_dengue.head(10)
```

```python
#Analisando os ultimos registros
df_dengue.tail(10)
```

```python
# Criação de coluna Ano
df_dengue['ano'] = df_dengue['data_infeccoes'].dt.year
```

```python
#del df_dengue['ano'] (Comando Python que deleta diretamente na base)
del df_dengue['ano']
```

```python
infeccoes_estado = df_dengue.groupby(['uf','ano'])['qtd_infeccoes'].sum().reset_index()
```

```python
infeccoes_estado.head()
```

## 3. Estatística Descritiva


```python
#Resumo dos dados 
infeccoes_estado.describe()
```

```python
plt.figure(figsize = (10,6))
sns.boxplot(data = infeccoes_estado, y = 'qtd_infeccoes')
```

```python
#Interpretação de estatistica descritiva utilizando o gráfico de Boxplot
plt.figure(figsize = (10,6))
sns.boxplot(data = infeccoes_estado, x = 'ano', y = 'qtd_infeccoes', palette='pastel')

#Personalizar
plt.title('quantidade de infeccoes por ano', fontsize = 14)
plt.xlabel('Ano', fontsize = 14)
plt.ylabel('Quantidade de infeccoes', fontsize = 14)
plt.grid(axis= 'y', linestyle = '--', alpha = 0.7)
plt.show ()
 
```

```python
#Filtros para Ano de 2023

infeccoes_estado_2023 = infeccoes_estado[infeccoes_estado['ano'] == 2023]
infeccoes_estado_2023

```

```python
#Entendendo o ano de 2023
infeccoes_estado_2023.describe()
```

```python
#gerando grafico para o Ano de 2023
plt.figure(figsize = (10,6))
sns.boxplot(data = infeccoes_estado_2023, y = 'qtd_infeccoes')
plt.title('boxplot de infeccoes por Estado / 2023')
plt.ylabel('Quantidade de infeccoes', fontsize = 14)
plt.grid(axis= 'y', linestyle = '--', alpha = 0.7)
plt.show ()
 
```

```python
plt.figure(figsize = (10,6))
plt.title('boxplot de infeccoes por Estado / 2023')
plt.ylabel('Quantidade de infeccoes', fontsize = 14)
plt.grid(axis= 'y', linestyle = '--', alpha = 0.7)
plt.show ()
```

```python
import plotly.express as px
fig = px.box(infeccoes_estado_2023, x = 'uf', y = 'qtd_infeccoes', title = 'boxplot', color = 'uf')
fig.show()
```

##Aula 4 - Graficos comparativos

```python
import seaborn as sns

plt.figure(figsize = (12,6))
sns.barplot(data = infeccoes_estado_2023, x = 'uf', y = 'qtd_infeccoes', palette = 'viridis')
plt.title ('Quantidade de infeccoes por Estado / 2023')
plt.xlabel('UF', fontsize = 14)
plt.ylabel('Quantidade de infeccoes', fontsize = 14)
plt.show()
```

```python
infeccoes_2023 = infeccoes_estado_2023.sort_values(by = 'qtd_infeccoes', ascending = False)
 
```

```python
import seaborn as sns

plt.figure(figsize = (12,6))
sns.barplot(data = infeccoes_2023, x = 'uf', y = 'qtd_infeccoes', palette = 'viridis')
plt.xticks (rotation = 40)
plt.title('Infeções por UF')
plt.grid(axis= 'y', linestyle = '--', alpha = 0.7)
plt.show ()
```

```python
#construindo um gráfico de tendencias
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize = (12,6))
ax = sns.regplot(data = infeccoes_estado, x = 'ano', y = 'qtd_infeccoes', scatter_kws = {"s": 15}, line_kws = {"color":"red"})

#personalizar o grafico
plt.title('Tendencia de infeccoes por ano', fontsize = '14')
plt.xlabel('Ano', fontsize = 14)
plt.ylabel('Quantidade de infecções', fontsize = 14)
ax.grid(True, linestyle = '--', alpha = 0.7)
plt.show()


```
